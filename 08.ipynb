````python
visits_com_coorte = visits.merge(primeira_visita, on='Uid')
visits_com_coorte['Semana_Atual'] = visits_com_coorte['Start Ts'].dt.to_period('W')

visits_com_coorte['Semanas_Desde_Primeira'] = (
    visits_com_coorte['Semana_Atual'] - visits_com_coorte['Primeira_Visita_Semana']
).apply(lambda x: x.n)

print("Dados com informação de coorte:")
print(visits_com_coorte[['Uid', 'Primeira_Visita_Semana', 'Semana_Atual', 'Semanas_Desde_Primeira']].head(10))
````

Identificador único de cada usuário
Exemplo: 16879256277535980062
2. Primeira_Visita_Semana:

A semana em que o usuário visitou pela primeira vez
Exemplo: 2017-12-18/2017-12-24 = usuário chegou nesta semana
3. Semana_Atual:

A semana da visita que estamos analisando
Pode ser a mesma da primeira visita ou semanas posteriores
4. Semanas_Desde_Primeira:

Quantas semanas se passaram desde a primeira visita
0 = primeira visita
1 = voltou na semana seguinte
10 = voltou 10 semanas depois

````python
def criar_tabela_retencao(df):
    
    tabela_retencao = df.groupby(['Primeira_Visita_Semana', 'Semanas_Desde_Primeira'])['Uid'].nunique().reset_index()
    tabela_retencao = tabela_retencao.pivot(index='Primeira_Visita_Semana', 
                                          columns='Semanas_Desde_Primeira', 
                                          values='Uid')
    
    
    for i in tabela_retencao.index:
        tabela_retencao.loc[i] = tabela_retencao.loc[i] / tabela_retencao.loc[i, 0]
    
    return tabela_retencao

tabela_retencao = criar_tabela_retencao(visits_com_coorte)
print("Tabela de Retenção (primeiras 5 coortes e 10 semanas):")
print(tabela_retencao.iloc[:5, :10])
````
